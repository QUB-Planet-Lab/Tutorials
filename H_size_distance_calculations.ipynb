```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import ipywidgets as widgets
from ipywidgets import Layout, Box, Text, FloatText, Label
```

```python
def apparent_mag(reduced_mag, r, delta):
    # calculate apparent magnitude
    apparent_mag=reduced_mag+5*np.log10(r*delta)
    return apparent_mag

def reduced_mag(apparent_mag, r, delta):
    # calculate apparent magnitude
    reduced_mag=apparent_mag-5*np.log10(r*delta)
    return reduced_mag

def calculate_radius(H, geometric_albedo):
    # taken from Harris  & Harris 1997 
    # Returns:
    #radius : float (in km)

    if geometric_albedo < 0:
        raise ValueError("Albedo must be greater than zero")

    diameter = (1329.0)/np.sqrt(geometric_albedo)*10.0**((-1.0)*H/5.0)
    return diameter/2.0

def plot_H_dist(mags,distances, ax, ylog=False, xlog=False):   
    
    for i in np.arange(len(mags)):
        H=reduced_mag(mags[i], distances, distances-1)
        ax.plot(distances, H, label=f"{mags[i]:.1f} mag")

    ax.set_xlabel("Heliocentric distance (au)",  fontsize=14)
    ax.set_ylabel("H",  fontsize=14)
    ax.tick_params(axis="both", labelsize=14)
    ax.legend(fontsize=14)
    ax.grid()

    if(xlog==True):
        ax.set_xscale("log") 
    
    if(ylog==True):
        ax.set_yscale("log") 


def plot_r_dist(mag,distances, geometric_albedo, ax,  ylog=False, xlog=False):   
    
    for i in np.arange(len(mags)):
        H=reduced_mag(mags[i], distances, distances-1)
        r=calculate_radius(H, geometric_albedo)
        ax.plot(distances, r, label=f"{mags[i]:.1f} mag")

    ax.set_xlabel("Heliocentric distance (au)",  fontsize=14)
    ax.set_ylabel("Radius (km)",  fontsize=14)
    ax.tick_params(axis="both", labelsize=14)
    ax.legend(fontsize=14)
    ax.grid()

    if(xlog==True):
        ax.set_xscale("log") 
    
    if(ylog==True):
        ax.set_yscale("log") 
```

```python
mags=[25.5, 25,24.5, 24.0, 23.5,23, 22.5, 22]
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 6*3))

distances=np.arange(21)
distances=distances[2:]

plot_H_dist(mags, distances, axes[0])

distances=np.arange(140)+10
plot_H_dist(mags, distances, axes[1])

distances=np.arange(151)
distances=distances[2:]
plot_H_dist(mags, distances, axes[2], xlog=True)
plt.tight_layout() 
plt.savefig('HvsDist.png')
```

```python

form_item_layout = Layout(
    display='flex',
    flex_flow='row',
    justify_content='space-between'
)



rBox=FloatText(value=30.00, step=1)
magBox=FloatText(value=25.00, step=0.1)

H=reduced_mag(magBox.value, rBox.value, rBox.value-1)
output=widgets.Output()

# set the initial value for the output box
with output:
    output.clear_output()
    H=reduced_mag(magBox.value, rBox.value, rBox.value-1)
    print(f'{H:.1f}')

def on_change(change):
    with output:
        output.clear_output()
        H=reduced_mag(magBox.value, rBox.value, rBox.value-1)
        print(f'{H:.1f}')

rBox.observe(on_change, names="value")
magBox.observe(on_change, names="value")


form_items = [
    Box([Label(value='Limiting Apparent Magnitude:'), 
         magBox], layout=form_item_layout),
    Box([Label(value='Heliocentric Distance (au):'), 
         rBox], layout=form_item_layout),
    Box([Label(value='Equivalent H:'), 
         output], layout=form_item_layout)
]

form = Box(form_items, layout=Layout(
    display='flex',
    flex_flow='column',
    border='solid 2px',
    align_items='stretch',
    width='80%'
))


form
```

```python

form_item_layout = Layout(
    display='flex',
    flex_flow='row',
    justify_content='space-between'
)



HBox=FloatText(value=5.00, step=0.5)
galbedoBox=FloatText(value=0.15, step=0.01)

output2=widgets.Output()

# set the initial value for the output box
with output2:
    output2.clear_output()
    radius=calculate_radius(HBox.value, galbedoBox.value)
    print(f'{radius:.2f}')

def on_change(change):
    with output2:
        output2.clear_output()
        radius=calculate_radius(HBox.value, galbedoBox.value)
        print(f'{radius:.2f}')

HBox.observe(on_change, names="value")
galbedoBox.observe(on_change, names="value")


form_items2 = [
    Box([Label(value='H:'), 
         HBox], layout=form_item_layout),
    Box([Label(value='Geometric Albedo:'), 
         galbedoBox], layout=form_item_layout),
    Box([Label(value='Radius (km):'), 
         output2], layout=form_item_layout)
]

form2 = Box(form_items2, layout=Layout(
    display='flex',
    flex_flow='column',
    border='solid 2px',
    align_items='stretch',
    width='80%'
))


form2
```
